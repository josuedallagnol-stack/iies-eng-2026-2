## 1. `if` — "Se isso for verdade, faça"

O bloco `if` executa um trecho de código **somente** quando a condição é verdadeira (`True`).  
Se for falsa (`False`), o bloco é ignorado e o programa segue em frente.

```python
if <condição>:
    # código executado apenas se a condição for True
```

> **Atenção ao recuo (indentação):** Python usa 4 espaços (ou 1 Tab) para delimitar o bloco.  
> Tudo que estiver recuado após o `if` pertence a ele.

````python
# Exemplo básico de if
temperatura = 35

if temperatura > 30:
    print("Está calor! Beba água.") #só executa se a condição do if for verdadeira

print("Fim do programa.")  # sempre executado
````

## 3. `elif` — "Ou então se..."

Quando há **mais de duas possibilidades**, encadeamos condições com `elif` (abreviação de *else if*).  
Python avalia de cima para baixo e executa **apenas o primeiro bloco** cuja condição for `True`.

```python
if <condição 1>:
    # executado se condição 1 for True
elif <condição 2>:
    # executado se condição 2 for True (e a 1 for False)
elif <condição 3>:
    # executado se condição 3 for True (e as anteriores forem False)
else:
    # executado se nenhuma condição acima for True
```

> Você pode ter **quantos `elif` quiser**, mas apenas **um `if`** e **um `else`** por estrutura.

````python
# Exemplo de if + elif + else — classificação por faixa etária
idade = int(input("Digite sua idade: "))

if idade < 10:
  print("faixa etaria 1")
elif idade < 50:
  print("faixa etaria 2")
elif idade < 80:
  print("faixa etaria 3")
else:
  print("faixa etaria 3")

````

**Ex 3.3** — Calculadora simples. Leia dois números e uma operação (`+`, `-`, `*`, `/`). Exiba o resultado.  
> 💡 Atenção: trate a divisão por zero!

````python
# Ex 3.3 — Resolva aqui

# seu código abaixo ↓
a = float(input("Digite o primeiro número: "))
b = float(input("Digite o segundo número: "))
operacao = input("Digite a operação (+, -, *, /): ")

if operacao == "+":
    print(f"Resultado: {a + b}")
elif operacao == "-":
    print(f"Resultado: {a - b}")
elif operacao == "*":
    print(f"Resultado: {a * b}")
elif operacao == "/":
    if b == 0:
        print("Erro: divisão por zero!")
    else:
        print(f"Resultado: {a / b}")
else:
    print("Operação inválida!")
````

Ex 3.7 — Leia o salário de um funcionário e calcule o desconto do IR:

Até R$2.000 → isento
R$2.001 a R$3.000 → 7,5%
R$3.001 a R$4.500 → 15%
Acima de R$4.500 → 22,5%

````python
salario = float(input("Digite o salário: "))

if salario <= 2000:
    aliquota = 0
elif salario <= 3000:
    aliquota = 7.5
elif salario <= 4500:
    aliquota = 15
else:
    aliquota = 22.5

desconto = salario * (aliquota / 100)
print(f"Alíquota: {aliquota}%")
print(f"Desconto: R$ {desconto:.2f}")
````
